```python
import os
import shutil
from icrawler.builtin import BingImageCrawler

# ── Configuration ────────────────────────────────────────────
NUM_IMAGES = 200          # images to download per class (increase for more data)
DATASET_DIR = "dataset"   # must match your notebook: os.listdir('dataset')

# Search keywords — more keywords = more variety = less overfitting
CLASSES = {
    "earphones": [
        "earphones product photo",
        "in-ear earphones white background",
        "wired earphones closeup",
        "wireless earphones buds",
        "earbuds product image",
    ],
    "headphones": [
        "headphones product photo",
        "over-ear headphones white background",
        "wireless headphones studio",
        "on-ear headphones closeup",
        "headset headphones product image",
    ],
    "phones": [
        "smartphone product photo",
        "mobile phone white background",
        "android phone closeup",
        "iphone product image",
        "cell phone front view",
    ],
}



def download_class(class_name, keywords, total_images, save_dir):
    """Download images for one class using multiple keywords."""
    os.makedirs(save_dir, exist_ok=True)

    images_per_keyword = max(1, total_images // len(keywords))
    print(f"\n{'='*55}")
    print(f"  Downloading: {class_name.upper()}")
    print(f"  Target : {total_images} images  |  Folder: {save_dir}")
    print(f"{'='*55}")

    for idx, keyword in enumerate(keywords):
        # Use a temp subfolder per keyword to avoid filename clashes
        temp_dir = os.path.join(save_dir, f"_temp_{idx}")
        os.makedirs(temp_dir, exist_ok=True)

        print(f"  [{idx+1}/{len(keywords)}] Searching: \"{keyword}\" ...")
        try:
            crawler = BingImageCrawler(storage={"root_dir": temp_dir})
            crawler.crawl(
                keyword=keyword,
                max_num=images_per_keyword,
                min_size=(100, 100),          # skip tiny/broken images
                file_idx_offset=0,
            )
        except Exception as e:
            print(f"    Warning: {e}")

        # Move downloaded files to main class folder with unique names
        existing = len([f for f in os.listdir(save_dir)
                        if os.path.isfile(os.path.join(save_dir, f))])
        for fname in os.listdir(temp_dir):
            src = os.path.join(temp_dir, fname)
            if os.path.isfile(src):
                ext = os.path.splitext(fname)[1].lower()
                if ext in (".jpg", ".jpeg", ".png", ".webp", ".bmp"):
                    dst = os.path.join(save_dir, f"{class_name}_{existing:05d}{ext}")
                    shutil.move(src, dst)
                    existing += 1

        shutil.rmtree(temp_dir, ignore_errors=True)

    count = len([f for f in os.listdir(save_dir)
                 if os.path.isfile(os.path.join(save_dir, f))])
    print(f"  ✓ {count} images saved in '{save_dir}'")
    return count


def main():
    print("\n" + "="*55)
    print("   EfficientNetB0 Custom Dataset Downloader")
    print("="*55)
    print(f"  Classes  : {list(CLASSES.keys())}")
    print(f"  Per class: {NUM_IMAGES} images")
    print(f"  Output   : ./{DATASET_DIR}/")
    print("="*55)

    totals = {}
    for class_name, keywords in CLASSES.items():
        save_dir = os.path.join(DATASET_DIR, class_name)
        count = download_class(class_name, keywords, NUM_IMAGES, save_dir)
        totals[class_name] = count

    print("\n" + "="*55)
    print("   DOWNLOAD COMPLETE — SUMMARY")
    print("="*55)
    grand_total = 0
    for cls, cnt in totals.items():
        print(f"  {cls:<15} : {cnt} images")
        grand_total += cnt
    print(f"  {'TOTAL':<15} : {grand_total} images")
    print("="*55)

    print("""
NEXT STEPS:
-----------
1. Open your notebook:  2-efficientnetB0_Custom_dataset.ipynb
2. Run all cells from the top.
3. The model will now train on your downloaded dataset.

TIPS TO REDUCE OVERFITTING:
-----------------------------
• Use at least 200+ images per class (increase NUM_IMAGES above).
• Add Data Augmentation in your notebook:
      from tensorflow.keras.preprocessing.image import ImageDataGenerator
      datagen = ImageDataGenerator(
          rotation_range=20,
          zoom_range=0.15,
          width_shift_range=0.2,
          height_shift_range=0.2,
          horizontal_flip=True,
      )
• Use Transfer Learning (change weights=None → weights='imagenet').
• Add Dropout layers after EfficientNet base.
""")


if __name__ == "__main__":
    main()
```
